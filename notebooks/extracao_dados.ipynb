## Extração das bases de dados para utilização do sistema:
A pasta de trabalho base do exemplo conta com 2 planilhas relevantes, sendo elas:

- ``Base_Referencia``

O notebook em questão se trata da coleta da base de refências dos lotes e dos dados do relatório para testes do sistema final.


### Extração da base de lotes:


```python
import pandas as pd

print("Iniciando busca pela base de dados ...")
# Leitura do dataset:
try:
    base_lotes_suja = pd.read_excel("../data/raw/inspecao_lotes_dia.xlsx", sheet_name="Base_Referencia", header=1)
except FileNotFoundError: 
    print("Não foi possivel encontrar o arquivo.")
except Exception as e:
    print(f"Algo ocorreu de errado ao encontrar o arquivo:/n{e}")

# Exportação do dataset limpo
if not base_lotes_suja.empty:
    print("Dataset encontrado, realizando limpeza")
    BASE_LOTES = base_lotes_suja.drop([23,24])
    BASE_LOTES.to_csv("../data/processed/base_lotes_referencia.csv", index=False)
    print("Dataset com a base dos lotes limpo e armazenado em 'data/processed/'/n/n")
    print(f"Head do dataset:\n{BASE_LOTES.head()}")

print("Fim do processo.")
```
